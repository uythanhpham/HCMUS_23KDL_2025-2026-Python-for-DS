# Class DataPreprocessor

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
from sklearn.ensemble import IsolationForest

class DataPreprocessor:
    """
    Lớp dùng để tiền xử lý dữ liệu, bao gồm đọc file, làm sạch, chuẩn hóa, 
    mã hóa biến phân loại và kỹ thuật đặc trưng.
    """
    
    # 1. Khởi tạo và Biểu diễn Đối tượng (__init__, __repr__)
  
    def __init__(self, data=None):
        """
        Khởi tạo DataPreprocessor.
        :param data: DataFrame ban đầu hoặc None.
        """
        if data is not None and not isinstance(data, pd.DataFrame):
            raise TypeError("Dữ liệu khởi tạo phải là một Pandas DataFrame.")
        self.df = data
        self.column_types = {}  # Lưu trữ kiểu dữ liệu đã phát hiện
        print("Dữ liệu đã được khởi tạo.")

    def __repr__(self):
        """
        Phương thức biểu diễn đối tượng một cách rõ ràng.
        """
        if self.df is not None:
            return f"DataPreprocessor(data_shape={self.df.shape}, columns={list(self.df.columns)})"
        return "DataPreprocessor(data=None)"

    # 2. Đọc Dữ liệu (Dùng @classmethod làm Alternative Constructor)
    
    @classmethod
    def read_data(cls, file_path, file_format, **kwargs):
        """
        Phương thức lớp để đọc dữ liệu từ nhiều định dạng khác nhau.
        :param file_path: Đường dẫn tới file.
        :param file_format: Định dạng file ('csv', 'xlsx', 'json').
        :param kwargs: Các tham số bổ sung cho hàm đọc của pandas (ví dụ: sep, sheet_name).
        :return: Một thực thể DataPreprocessor với dữ liệu đã được nạp.
        """
        try:
            if file_format == 'csv':
                df = pd.read_csv(file_path, **kwargs)
            elif file_format == 'xlsx':
                df = pd.read_excel(file_path, **kwargs)
            elif file_format == 'json':
                df = pd.read_json(file_path, **kwargs)
            else:
                raise ValueError(f"Định dạng file '{file_format}' không được hỗ trợ.")
            
            # Trả về một biến mới sau khi đọc thành công
            return cls(data=df)
        
        except FileNotFoundError:
            print(f"LỖI: Không tìm thấy file tại đường dẫn: {file_path}")
            return cls(data=None) # Trả về rỗng
        except Exception as e:
            print(f"LỖI XỬ LÝ DỮ LIỆU: {e}")
            return cls(data=None)


    # 3. Phát hiện Kiểu Dữ liệu (Tự động hóa)

    def detect_data_types(self):
        """
        Tự động phát hiện và phân loại các cột thành số, phân loại, hoặc datetime.
        """
        if self.df is None:
            print("Chưa có dữ liệu để phát hiện kiểu.")
            return

        self.column_types = {
            'numerical': [],
            'categorical': [],
            'datetime': []
        }
        
        for col in self.df.columns:
            # Kiểm tra Datetime
            if pd.api.types.is_datetime64_any_dtype(self.df[col]):
                self.column_types['datetime'].append(col)
            # Kiểm tra Số (Số nguyên hoặc số thực)
            elif pd.api.types.is_numeric_dtype(self.df[col]):
                self.column_types['numerical'].append(col)
            # Mọi thứ còn lại coi là Phân loại (Object, String)
            else:
                self.column_types['categorical'].append(col)
        
        print(f"Phát hiện kiểu dữ liệu hoàn tất: {self.column_types}")

    # 4. Xử lý Dữ liệu Thiếu (Missing Data)
    
    def handle_missing_values(self, columns, strategy='mean'):
        """
        Điền giá trị thiếu (NaN) theo chiến lược đã chọn.
        :param columns: Danh sách các cột cần xử lý.
        :param strategy: 'mean', 'median', 'mode', 'ffill', 'bfill', hoặc một giá trị cụ thể.
        """
        if self.df is None: return

        try:
            for col in columns:
                if strategy in ['mean', 'median', 'mode']:
                    # Áp dụng cho columns dữ liệu số
                    if col in self.column_types.get('numerical', []):
                        fill_value = self.df[col].agg(strategy)
                        self.df[col].fillna(fill_value, inplace=True)
                    # Áp dụng cho columns phân loại (chỉ dùng mode)
                    elif strategy == 'mode':
                         fill_value = self.df[col].mode()[0]
                         self.df[col].fillna(fill_value, inplace=True)
                
                elif strategy in ['ffill', 'bfill']:
                    self.df[col].fillna(method=strategy, inplace=True)
                    
                elif isinstance(strategy, (int, float, str)):
                    # Điền bằng giá trị cụ thể
                    self.df[col].fillna(strategy, inplace=True)
                else:
                    raise ValueError("Chiến lược điền khuyết không hợp lệ.")
            
            print(f"Xử lý giá trị thiếu cho {columns} với chiến lược '{strategy}' hoàn tất.")
        except Exception as e:
            print(f"LỖI Xử lý giá trị thiếu: {e}")

    # 5. Phát hiện và Xử lý Ngoại Lai (Outliers)

    def handle_outliers(self, column, method='iqr', threshold=1.5):
        """
        Phát hiện và loại bỏ/thay thế giá trị ngoại lai.
        :param column: Cột số cần kiểm tra.
        :param method: 'iqr', 'zscore', 'isolation_forest'.
        :param threshold: Ngưỡng cho IQR/Z-score.
        """
        if self.df is None or column not in self.df.columns or column not in self.column_types.get('numerical', []): 
            print("Dữ liệu hoặc cột không hợp lệ.")
            return
            
        series = self.df[column].copy()
        outlier_indices = np.array([])

        if method == 'iqr':
            Q1 = series.quantile(0.25)
            Q3 = series.quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - threshold * IQR
            upper_bound = Q3 + threshold * IQR
            outlier_indices = series[(series < lower_bound) | (series > upper_bound)].index
            
        elif method == 'zscore':
            z_scores = (series - series.mean()) / series.std()
            outlier_indices = series[abs(z_scores) > threshold].index
            
        elif method == 'isolation_forest':
            model = IsolationForest(contamination='auto', random_state=42)
            # -1 là ngoại lai, 1 là bình thường
            outlier_mask = model.fit_predict(series.values.reshape(-1, 1)) == -1
            outlier_indices = series[outlier_mask].index

        # Loại bỏ các hàng chứa ngoại lai (ví dụ đơn giản)
        self.df.drop(outlier_indices, inplace=True)
        print(f"Đã loại bỏ {len(outlier_indices)} ngoại lai trong cột '{column}' bằng phương pháp '{method}'.")


    # 6. Chuẩn hóa/Tỷ lệ hóa (Scaling)
    
    def scale_data(self, columns, scaler_type='standard'):
        """
        Chuẩn hóa dữ liệu bằng StandardScaler, MinMaxScaler hoặc hàm tự cài đặt.
        :param columns: Danh sách các cột số cần chuẩn hóa.
        :param scaler_type: 'standard', 'minmax', 'custom_minmax'.
        """
        if self.df is None: return

        if scaler_type == 'standard':
            scaler = StandardScaler()
        elif scaler_type == 'minmax':
            scaler = MinMaxScaler()
        elif scaler_type == 'custom_minmax':
            # Tự cài đặt Min-Max Scaling (dùng @staticmethod)
            self.df[columns] = self._custom_minmax_scaler(self.df[columns])
            print("Chuẩn hóa Min-Max (Custom) hoàn tất.")
            return
        else:
            raise ValueError("Loại scaler không hợp lệ.")
            
        self.df[columns] = scaler.fit_transform(self.df[columns])
        print(f"Chuẩn hóa '{scaler_type}' cho {columns} hoàn tất.")

    @staticmethod
    def _custom_minmax_scaler(data):
        """
        Hàm tiện ích tĩnh (không cần self hay cls) để tính toán Min-Max Scaling.
        """
        return (data - data.min()) / (data.max() - data.min())

    # 7. Mã hóa Biến Phân loại (Encoding)
    
    def encode_categorical(self, column, encoder_type='onehot'):
        """
        Mã hóa biến phân loại.
        :param column: Tên cột cần mã hóa.
        :param encoder_type: 'onehot', 'label', 'custom_ordinal'.
        """
        if self.df is None or column not in self.df.columns: return

        if encoder_type == 'onehot':
            encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
            encoded_data = encoder.fit_transform(self.df[[column]])
            
            # Tạo DataFrame mới từ kết quả mã hóa và nối vào df gốc
            feature_names = encoder.get_feature_names_out([column])
            encoded_df = pd.DataFrame(encoded_data, columns=feature_names, index=self.df.index)
            self.df = pd.concat([self.df.drop(column, axis=1), encoded_df], axis=1)
            
        elif encoder_type == 'label':
            encoder = LabelEncoder()
            self.df[column] = encoder.fit_transform(self.df[column].astype(str))
            
        elif encoder_type == 'custom_ordinal':
            # Mã hóa dựa trên ánh xạ tự định nghĩa
            mapping = self._get_custom_mapping(column)
            self.df[column] = self.df[column].map(mapping)
            
        else:
            raise ValueError("Loại encoder không hợp lệ.")
            
        print(f"Mã hóa '{encoder_type}' cho cột '{column}' hoàn tất.")

    @staticmethod
    def _get_custom_mapping(column_name):
        """
        Hàm tiện ích tĩnh để trả về ánh xạ dạng text sang số. 
        Trong thực tế, ánh xạ này sẽ được định nghĩa dựa trên logic nghiệp vụ.
        """
        # Ví dụ: Mức độ thỏa mãn
        if column_name == 'satisfaction':
            return {'Low': 0, 'Medium': 1, 'High': 2}
        
        # Ví dụ: Giới tính
        if column_name == 'gender':
            return {'Male': 0, 'Female': 1}
        
        return {} # Trả về rỗng nếu không tìm thấy ánh xạ


    # 8. Kỹ thuật Đặc trưng (Feature Engineering)

    def datetime_feature_engineering(self, datetime_column, features=['year', 'month', 'dayofweek']):
        """
        Tạo các đặc trưng mới từ cột ngày giờ.
        :param datetime_column: Cột chứa dữ liệu datetime.
        :param features: Danh sách các đặc trưng cần tạo ('year', 'month', 'day', 'dayofweek', 'hour').
        """
        if self.df is None or datetime_column not in self.df.columns: return

        try:
            # Đảm bảo cột ở định dạng datetime
            self.df[datetime_column] = pd.to_datetime(self.df[datetime_column])
            
            for feat in features:
                if feat == 'year':
                    self.df[f'{datetime_column}_year'] = self.df[datetime_column].dt.year
                elif feat == 'month':
                    self.df[f'{datetime_column}_month'] = self.df[datetime_column].dt.month
                elif feat == 'day':
                    self.df[f'{datetime_column}_day'] = self.df[datetime_column].dt.day
                elif feat == 'dayofweek':
                    self.df[f'{datetime_column}_dayofweek'] = self.df[datetime_column].dt.dayofweek
                elif feat == 'hour':
                    self.df[f'{datetime_column}_hour'] = self.df[datetime_column].dt.hour
            
            print(f"Kỹ thuật đặc trưng datetime cho cột '{datetime_column}' hoàn tất.")
        except Exception as e:
            print(f"LỖI Kỹ thuật đặc trưng Datetime: {e}")

    # 9. Ghi Dữ liệu & Trả về DataFrame

    def write_processed_data(self, file_path, file_format='csv'):
        """
        Ghi DataFrame đã xử lý ra file mới.
        """
        if self.df is None:
            print("Không có dữ liệu để ghi.")
            return

        try:
            if file_format == 'csv':
                self.df.to_csv(file_path, index=False)
            elif file_format == 'xlsx':
                self.df.to_excel(file_path, index=False)
            elif file_format == 'json':
                self.df.to_json(file_path, orient='records')
            else:
                raise ValueError("Định dạng ghi file không được hỗ trợ.")
            print(f"Dữ liệu đã được ghi thành công ra file: {file_path}")
        except Exception as e:
            print(f"LỖI khi ghi dữ liệu: {e}")

    def get_processed_data(self):
        """
        Trả về DataFrame đã sẵn sàng cho mô hình.
        """
        return self.df
```

## Ví dụ áp dụng:

### Tạo Data giả sử

```python
# 1. Tạo DataFrame mẫu (Thực tế, bạn sẽ dùng read_data để đọc file)
data = {
    'Giới_Tính': ['Male', 'Female', 'Female', 'Male', 'Male', 'Female', 'Female', 'Male', 'Male', 'Female'],
    'Độ_Tuổi': [25, 30, 45, 50, 22, np.nan, 35, 28, 60, 32],
    'Lương': [50000, 60000, 75000, 80000, 52000, 65000, 70000, 1000000, 48000, np.nan], # 1,000,000 là ngoại lai
    'Thỏa_Mãn': ['Medium', 'High', 'Low', 'Medium', 'Low', 'High', 'Medium', 'High', 'Medium', 'Low'],
    'Ngày_Tham_Gia': ['2022-01-15', '2020-05-20', '2023-11-01', '2019-03-10', '2022-08-25', '2021-02-01', '2023-01-01', '2020-10-10', '2018-04-04', '2021-07-07']
}
df_raw = pd.DataFrame(data)

# Tạo một file CSV ảo để dùng read_data()
df_raw.to_csv("employee_data.csv", index=False) 

# In ra 5 dòng đầu của dữ liệu thô
print("--- Dữ liệu thô ban đầu ---")
print(df_raw.head())
print("\n")
```

### Áp dụng Class

```python
# --- Bắt đầu quá trình tiền xử lý ---

# 1. Đọc dữ liệu (Sử dụng @classmethod)
# dp là đối tượng DataPreprocessor mới được tạo
dp = DataPreprocessor.read_data("employee_data.csv", file_format='csv')
print(f"Đối tượng sau khi đọc: {dp}")

# 2. Phát hiện kiểu dữ liệu
dp.detect_data_types()

# --- BƯỚC 1: LÀM SẠCH DỮ LIỆU ---
print("\n--- BƯỚC 1: LÀM SẠCH DỮ LIỆU ---")

# 3. Xử lý giá trị thiếu:
# a. Cột 'Lương' (Numerical) dùng Median (ít bị ảnh hưởng bởi ngoại lai hơn Mean)
dp.handle_missing_values(columns=['Lương'], strategy='median')
# b. Cột 'Độ_Tuổi' (Numerical) dùng Mean
dp.handle_missing_values(columns=['Độ_Tuổi'], strategy='mean')

# 4. Xử lý Ngoại Lai: Cột 'Lương' dùng phương pháp IQR (Ngưỡng 3.0 để giữ lại hầu hết dữ liệu)
dp.handle_outliers(column='Lương', method='iqr', threshold=3.0)

# --- BƯỚC 2: KỸ THUẬT ĐẶC TRƯNG ---
print("\n--- BƯỚC 2: KỸ THUẬT ĐẶC TRƯNG ---")

# 5. Kỹ thuật Datetime: Tách 'Ngày_Tham_Gia' thành năm và tháng
dp.datetime_feature_engineering(datetime_column='Ngày_Tham_Gia', features=['year', 'month'])
# Loại bỏ cột ngày tháng gốc sau khi tách
dp.df.drop(columns=['Ngày_Tham_Gia'], inplace=True)

# --- BƯỚC 3: CHUYỂN ĐỔI DỮ LIỆU ---
print("\n--- BƯỚC 3: CHUYỂN ĐỔI DỮ LIỆU ---")

# 6. Mã hóa Biến Phân loại:
# a. Biến 'Giới_Tính' (Nominal) dùng OneHotEncoder
dp.encode_categorical(column='Giới_Tính', encoder_type='onehot')

# b. Biến 'Thỏa_Mãn' (Ordinal) dùng Custom Mapping (đã định nghĩa trong Class)
dp.encode_categorical(column='Thỏa_Mãn', encoder_type='custom_ordinal')
dp.df.rename(columns={'Thỏa_Mãn': 'Thỏa_Mãn_Ordinal'}, inplace=True) # Đổi tên cho rõ ràng

# 7. Chuẩn hóa Dữ liệu Số: 'Độ_Tuổi' và 'Lương' dùng Standard Scaler
dp.scale_data(columns=['Độ_Tuổi', 'Lương'], scaler_type='standard')

# --- KẾT QUẢ ---

# 8. Trả về DataFrame đã xử lý
df_processed = dp.get_processed_data()
```
